# **Memahami kondisi awal dataset**

```python
print("=== INFORMASI DATASET ===")
df.info()

print("\n=== UKURAN DATASET ===")
print(df.shape)

print("\n=== STATISTIK AWAL ===")
display(df.describe(include="all").T)

print("\n=== MISSING VALUES ===")
display(df.isnull().sum().to_frame("Jumlah Missing"))

print("\n=== DATA DUPLIKAT ===")
print("Duplikat seluruh baris:", df.duplicated().sum())
print("Duplikat ID:", df["id"].duplicated().sum())
```

```python
# Membuat salinan dataset
dfClean = df.copy()

# 1. Mengisi missing values pada kolom kategorikal
dfClean["product_name"] = dfClean["product_name"].fillna("Unknown")
dfClean["product_brand"] = dfClean["product_brand"].fillna("Unknown")

# 2. Mengubah kolom tanggal menjadi datetime
dfClean["created_at"] = pd.to_datetime(
    dfClean["created_at"],
    errors="coerce",
    format="mixed"
)

dfClean["sold_at"] = pd.to_datetime(
    dfClean["sold_at"],
    errors="coerce",
    format="mixed"
)

# 3. Menghapus duplikasi baris
jumlahSebelum = len(dfClean)

dfClean = dfClean.drop_duplicates()
dfClean = dfClean.drop_duplicates(subset=["id"])

jumlahSesudah = len(dfClean)

print("Jumlah data sebelum:", jumlahSebelum)
print("Jumlah data sesudah:", jumlahSesudah)
print("Jumlah duplikat dihapus:", jumlahSebelum - jumlahSesudah)

# 4. Menambahkan status penjualan
dfClean["sales_status"] = np.where(
    dfClean["sold_at"].notna(),
    "Terjual",
    "Belum Terjual"
)

# 5. Mengecek missing values setelah preprocessing
print("\n=== MISSING VALUES SETELAH PREPROCESSING ===")
display(dfClean.isnull().sum().to_frame("Missing Values"))

print("\n=== TIPE DATA ===")
dfClean.info()
```

# **Mengidentifikasi outlier menggunakan metode IQR**

```python
# Kolom yang akan diperiksa
numericColumns = ["cost", "product_retail_price"]

outlierSummary = []

for column in numericColumns:
    q1 = dfClean[column].quantile(0.25)
    q3 = dfClean[column].quantile(0.75)
    iqr = q3 - q1

    lowerBound = q1 - 1.5 * iqr
    upperBound = q3 + 1.5 * iqr

    outlierMask = (
        (dfClean[column] < lowerBound) |
        (dfClean[column] > upperBound)
    )

    outlierSummary.append({
        "Variabel": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Batas Bawah": lowerBound,
        "Batas Atas": upperBound,
        "Jumlah Outlier": outlierMask.sum(),
        "Persentase (%)": outlierMask.mean() * 100
    })

outlierSummary = pd.DataFrame(outlierSummary)

display(outlierSummary.round(2))
```

# **Menangani outlier**

```python
dfClean["is_outlier"] = False

for column in numericColumns:
    q1 = dfClean[column].quantile(0.25)
    q3 = dfClean[column].quantile(0.75)
    iqr = q3 - q1

    lowerBound = q1 - 1.5 * iqr
    upperBound = q3 + 1.5 * iqr

    outlierMask = (
        (dfClean[column] < lowerBound) |
        (dfClean[column] > upperBound)
    )

    dfClean["is_outlier"] |= outlierMask

    # Kolom tambahan untuk analisis tanpa nilai ekstrem
    dfClean[column + "_capped"] = dfClean[column].clip(
        lower=lowerBound,
        upper=upperBound
    )

print("Total baris terindikasi outlier:",
      dfClean["is_outlier"].sum())

print("Total baris normal:",
      (~dfClean["is_outlier"]).sum())

print("Jumlah keseluruhan:", len(dfClean))

display(
    dfClean[
        ["cost", "cost_capped",
         "product_retail_price",
         "product_retail_price_capped",
         "is_outlier"]
    ].head(10)
)
```

# **Feature engineering**

```python
# 1. Menghitung estimasi laba kotor
dfClean["gross_profit"] = (
    dfClean["product_retail_price"] - dfClean["cost"]
)

# 2. Menghitung profit margin (%)
dfClean["profit_margin"] = np.where(
    dfClean["product_retail_price"] > 0,
    dfClean["gross_profit"] /
    dfClean["product_retail_price"] * 100,
    np.nan
)

# 3. Menghitung markup (%)
dfClean["markup_percentage"] = np.where(
    dfClean["cost"] > 0,
    dfClean["gross_profit"] /
    dfClean["cost"] * 100,
    np.nan
)

# 4. Tahun pencatatan item
dfClean["created_year"] = dfClean["created_at"].dt.year

# 5. Tahun penjualan
dfClean["sold_year"] = dfClean["sold_at"].dt.year

# 6. Periode penjualan bulanan
dfClean["sold_month"] = (
    dfClean["sold_at"].dt.to_period("M")
)

# 7. Durasi penjualan dalam hari
dfClean["days_to_sell"] = (
    dfClean["sold_at"] - dfClean["created_at"]
).dt.total_seconds() / 86400

# 8. Menandai durasi yang tidak valid
dfClean["invalid_sale_duration"] = (
    dfClean["days_to_sell"] < 0
)

dfClean.loc[
    dfClean["invalid_sale_duration"],
    "days_to_sell"
] = np.nan

print("Feature engineering berhasil!")

display(
    dfClean[
        ["cost", "product_retail_price", "gross_profit",
         "profit_margin", "markup_percentage",
         "sales_status", "days_to_sell"]
    ].head(10)
)
```

# **Mengelompokkan produk berdasarkan segmen harga**

```python
# Segmentasi menggunakan ranking agar jumlah kategori stabil
priceRank = dfClean["product_retail_price"].rank(
    method="first"
)

dfClean["price_segment"] = pd.qcut(
    priceRank,
    q=4,
    labels=["Low", "Medium", "High", "Premium"]
)

print("Distribusi segmen harga:")
display(dfClean["price_segment"].value_counts().sort_index())
```

# **Analisis estimasi keuntungan setiap kategori**

```python
dfSold = dfClean[
    dfClean["sales_status"] == "Terjual"
].copy()

categoryProfit = dfSold.groupby(
    "product_category"
).agg(
    total_items=("id", "count"),
    total_revenue=("product_retail_price", "sum"),
    total_cost=("cost", "sum"),
    total_profit=("gross_profit", "sum")
).reset_index()

categoryProfit["profit_margin"] = np.where(
    categoryProfit["total_revenue"] > 0,
    categoryProfit["total_profit"] /
    categoryProfit["total_revenue"] * 100,
    np.nan
)

categoryProfit = categoryProfit.sort_values(
    by="total_profit",
    ascending=False
)

print("=== ESTIMASI KEUNTUNGAN PER KATEGORI ===")
display(categoryProfit.round(2))
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=categoryProfit.head(10),
    x="total_profit",
    y="product_category",
    color="seagreen"
)

plt.title("Top 10 Kategori Berdasarkan Estimasi Laba Kotor")
plt.xlabel("Total Estimasi Laba Kotor")
plt.ylabel("Kategori Produk")
plt.tight_layout()
plt.show()
```

# **Analisis tren penjualan bulanan**

```python
monthlySales = (
    dfSold.groupby("sold_month", observed=True)
    .agg(
        total_sold=("id", "count"),
        total_revenue=("product_retail_price", "sum"),
        total_profit=("gross_profit", "sum")
    )
    .reset_index()
)

monthlySales["sold_month"] = (
    monthlySales["sold_month"].dt.to_timestamp()
)

monthlySales = monthlySales.sort_values("sold_month")

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthlySales,
    x="sold_month",
    y="total_sold",
    marker="o",
    color="royalblue"
)

plt.title("Tren Jumlah Item Terjual per Bulan")
plt.xlabel("Periode Penjualan")
plt.ylabel("Jumlah Item Terjual")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

display(monthlySales.head(10))
```
